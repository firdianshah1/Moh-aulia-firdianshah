```
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Baca gambar grayscale
img = cv2.imread('unta.jpeg', 0)

# 2. Ambil satu baris (misal baris ke-100) sebagai sinyal 1D
row = img[100, :]   # sinyal spasial 1D

# 3. Lakukan Fourier Transform 1D
F = np.fft.fft(row)
Fshift = np.fft.fftshift(F)

# 4. Hitung magnitude spectrum
magnitude = np.abs(Fshift)
freq = np.fft.fftfreq(len(row))

# 5. Tampilkan hasil
plt.figure(figsize=(12,6))


plt.subplot(2,1,1)
plt.plot(row)
plt.title('Sinyal Asli (Domain Spasial)')
plt.xlabel('Posisi Piksel')
plt.ylabel('Intensitas')

plt.subplot(2,1,2)
plt.plot(freq, magnitude)
plt.title('Hasil Fourier 1D (Domain Frekuensi)')
plt.xlabel('Frekuensi (u)')
plt.ylabel('|F(u)|')
plt.tight_layout()
plt.show()
```

```
# ============================================
# Hilangkan noise di domain frekuensi (2D FFT)
# ============================================

import cv2
import numpy as np
import matplotlib.pyplot as plt

# ---- 1. Baca gambar grayscale ----
# (kamu bisa ganti 'lena.jpg' dengan file lain, atau upload di Colab)
!wget -q https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png -O lena.png
img = cv2.imread('unta.jpeg', cv2.IMREAD_GRAYSCALE)

# ---- 2. Tambahkan noise acak ----
noisy = img + 30 * np.random.randn(*img.shape)
noisy = np.clip(noisy, 0, 255).astype(np.uint8)

# ---- 3. Transformasi Fourier 2D ----
f = np.fft.fft2(noisy)
fshift = np.fft.fftshift(f)
magnitude_spectrum = 20 * np.log(np.abs(fshift) + 1)

# ---- 4. Buat filter low-pass (mask bundar) ----
rows, cols = noisy.shape
crow, ccol = rows // 2, cols // 2  # titik tengah
mask = np.zeros((rows, cols), np.uint8)
r = 50  # radius (semakin besar = lebih blur, tapi noise hilang)
cv2.circle(mask, (ccol, crow), r, 1, thickness=-1)

# ---- 5. Terapkan mask di domain frekuensi ----
fshift_filtered = fshift * mask

# ---- 6. Transformasi balik ke domain spasial ----
f_ishift = np.fft.ifftshift(fshift_filtered)
img_filtered = np.fft.ifft2(f_ishift)
img_filtered = np.abs(img_filtered)

# ---- 7. Tampilkan hasil ----
plt.figure(figsize=(12,6))

plt.subplot(1,3,1)
plt.imshow(noisy, cmap='gray')
plt.title('Gambar Ber-noise')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(np.log(1+np.abs(fshift)), cmap='gray')
plt.title('Spektrum Frekuensi (FFT)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_filtered, cmap='gray')
plt.title('Setelah Filter Frekuensi (Noise Hilang)')
plt.axis('off')

plt.show()
```

```
# =========================================
# Hilangkan noise dengan Low-Pass Filter FFT
# =========================================

import numpy as np
import cv2
import matplotlib.pyplot as plt

# --- 1. Ambil contoh gambar (otomatis download) ---
!wget -q https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png -O lena.png

# Baca gambar grayscale
img = cv2.imread('unta.jpeg', cv2.IMREAD_GRAYSCALE)

# --- 2. Tambahkan noise acak (Gaussian noise) ---
noisy = img + 30 * np.random.randn(*img.shape)
noisy = np.clip(noisy, 0, 255).astype(np.uint8)

# --- 3. Ubah ke domain frekuensi (FFT 2D) ---
f = np.fft.fft2(noisy)
fshift = np.fft.fftshift(f)

# --- 4. Buat Low-Pass Filter (lingkaran di tengah) ---
rows, cols = noisy.shape
crow, ccol = rows // 2, cols // 2
r = 50  # radius filter (atur sesuai kebutuhan)

mask = np.zeros((rows, cols), np.uint8)
cv2.circle(mask, (ccol, crow), r, 1, thickness=-1)

# --- 5. Terapkan filter ---
fshift_filtered = fshift * mask

# --- 6. Transformasi balik ke domain spasial (IFFT) ---
f_ishift = np.fft.ifftshift(fshift_filtered)
img_filtered = np.fft.ifft2(f_ishift)
img_filtered = np.abs(img_filtered)

# --- 7. Tampilkan hasil ---
plt.figure(figsize=(14,6))

plt.subplot(1,3,1)
plt.imshow(noisy, cmap='gray')
plt.title('Gambar Ber-Noise')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(np.log(1 + np.abs(fshift)), cmap='gray')
plt.title('Spektrum Frekuensi (FFT)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_filtered, cmap='gray')
plt.title('Setelah Low-Pass Filter')
plt.axis('off')

plt.show()
```

```
# ==========================================================
# Menghilangkan noise di domain frekuensi dengan High-Pass Filter
# ==========================================================

import numpy as np
import cv2
import matplotlib.pyplot as plt

# --- 1. Ambil contoh gambar ---
!wget -q https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png -O lena.png

# Baca gambar dalam grayscale
img = cv2.imread('unta.jpeg', cv2.IMREAD_GRAYSCALE)

# --- 2. Tambahkan noise acak (Gaussian noise) ---
noisy = img + 25 * np.random.randn(*img.shape)
noisy = np.clip(noisy, 0, 255).astype(np.uint8)

# --- 3. Ubah ke domain frekuensi (FFT 2D) ---
f = np.fft.fft2(noisy)
fshift = np.fft.fftshift(f)

# --- 4. Buat High-Pass Filter (mask dengan area tengah dihapus) ---
rows, cols = noisy.shape
crow, ccol = rows // 2, cols // 2
r = 30  # radius area frekuensi rendah yang akan dihilangkan

mask = np.ones((rows, cols), np.uint8)
cv2.circle(mask, (ccol, crow), r, 0, thickness=-1)  # nolkan area tengah (low freq)

# --- 5. Terapkan filter di domain frekuensi ---
fshift_filtered = fshift * mask

# --- 6. Kembalikan ke domain spasial (IFFT) ---
f_ishift = np.fft.ifftshift(fshift_filtered)
img_filtered = np.fft.ifft2(f_ishift)
img_filtered = np.abs(img_filtered)

# --- 7. Tampilkan hasil ---
plt.figure(figsize=(14,6))

plt.subplot(1,3,1)
plt.imshow(noisy, cmap='gray')
plt.title('Gambar Ber-Noise')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(np.log(1 + np.abs(fshift)), cmap='gray')
plt.title('Spektrum Frekuensi (FFT)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_filtered, cmap='gray')
plt.title('Setelah High-Pass Filter')
plt.axis('off')

plt.show()
```

```
# ==========================================================
# Menghilangkan noise periodik di domain frekuensi (Notch Filter)
# ==========================================================

import numpy as np
import cv2
import matplotlib.pyplot as plt

# --- 1. Unduh dan baca gambar contoh ---
!wget -q https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png -O lena.png
img = cv2.imread('unta.jpeg', cv2.IMREAD_GRAYSCALE)

# --- 2. Tambahkan noise periodik (simulasi) ---
rows, cols = img.shape
X, Y = np.meshgrid(np.arange(cols), np.arange(rows))
periodic_noise = 20 * np.sin(0.2 * X + 0.3 * Y)  # noise sinusoidal
noisy = np.clip(img + periodic_noise, 0, 255).astype(np.uint8)

# --- 3. FFT 2D ---
f = np.fft.fft2(noisy)
fshift = np.fft.fftshift(f)
magnitude_spectrum = np.log(1 + np.abs(fshift))

# --- 4. Buat Notch Reject Filter (dua titik simetris) ---
mask = np.ones((rows, cols), np.uint8)

# posisi frekuensi noise (ditentukan manual dari spektrum)
crow, ccol = rows // 2, cols // 2
notch_radius = 10

# lokasi notch kiri dan kanan (disesuaikan tergantung noise)
offset = 40
cv2.circle(mask, (ccol + offset, crow + offset), notch_radius, 0, -1)
cv2.circle(mask, (ccol - offset, crow - offset), notch_radius, 0, -1)

# --- 5. Terapkan Notch Filter ---
fshift_filtered = fshift * mask

# --- 6. Inverse FFT ---
f_ishift = np.fft.ifftshift(fshift_filtered)
img_filtered = np.fft.ifft2(f_ishift)
img_filtered = np.abs(img_filtered)

# --- 7. Tampilkan hasil ---
plt.figure(figsize=(14,7))

plt.subplot(1,3,1)
plt.imshow(noisy, cmap='gray')
plt.title('Gambar Ber-noise Periodik')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(np.log(1 + np.abs(fshift)), cmap='gray')
plt.title('Spektrum Frekuensi (Tampak Titik Noise)')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_filtered, cmap='gray')
plt.title('Setelah Notch Filter')
plt.axis('off')

plt.show()
```

```
# ==========================================================
# Menghilangkan noise di domain frekuensi dengan Median Filter
# ==========================================================

import numpy as np
import cv2
import matplotlib.pyplot as plt
from scipy.signal import medfilt2d  # median filter 2D

# --- 1. Ambil contoh gambar ---
!wget -q https://upload.wikimedia.org/wikipedia/en/7/7d/Lenna_%28test_image%29.png -O lena.png
img = cv2.imread('unta.jpeg', cv2.IMREAD_GRAYSCALE)

# --- 2. Tambahkan noise impulsif (salt & pepper noise) ---
noisy = img.copy()
prob = 0.02  # proporsi noise
rand = np.random.rand(*img.shape)
noisy[rand < prob/2] = 0
noisy[rand > 1 - prob/2] = 255

# --- 3. FFT 2D ---
f = np.fft.fft2(noisy)
fshift = np.fft.fftshift(f)
magnitude = np.log(1 + np.abs(fshift))

# --- 4. Terapkan Median Filter pada spektrum magnitude ---
magnitude_filtered = medfilt2d(magnitude, kernel_size=5)

# --- 5. Rekonstruksi kembali amplitudo hasil median ---
# Gunakan phase asli dari FFT agar tidak kehilangan informasi
phase = np.angle(fshift)
fshift_filtered = np.exp(magnitude_filtered - np.log(1)) * np.exp(1j * phase)

# --- 6. Inverse FFT untuk kembali ke domain spasial ---
f_ishift = np.fft.ifftshift(fshift)
img_back = np.fft.ifft2(f_ishift)
img_back = np.abs(img_back)

f_ishift2 = np.fft.ifftshift(fshift_filtered)
img_filtered = np.fft.ifft2(f_ishift2)
img_filtered = np.abs(img_filtered)

# --- 7. Tampilkan hasil ---
plt.figure(figsize=(14,7))

plt.subplot(1,3,1)
plt.imshow(noisy, cmap='gray')
plt.title('Gambar dengan Noise (Salt & Pepper)')
plt.axis('off')

plt.subplot(1,3,2)
plt.imshow(magnitude, cmap='gray')
plt.title('Spektrum Frekuensi Asli')
plt.axis('off')

plt.subplot(1,3,3)
plt.imshow(img_filtered, cmap='gray')
plt.title('Setelah Median Filter di Domain Frekuensi')
plt.axis('off')

plt.show()
```
